*Exercise 1:* Write a function that creates an **n×n matrix** (a list of lists) representing the state of a tic‑tac‑toe game.

Use the integers:

- `0` = empty
- `1` = `"X"`
- `2` = `"O"`


````python
# Write your solution here
def create_board(n):
    board = []  # make an empty board

    for i in range(n):  # for each row
        row = []  # make an empty row
        for j in range(n):  # for each column
            row.append(0)  # put 0 in the row (empty space)
        board.append(row)  # add the row to the board

    return board  # give back the board
````

````python
# Write your solution here
def draw_tic_tac_toe(board):
    for i in range(len(board)):  # for each row
        print(" ---" * len(board[i]))  # top line

        print("|", end="")  
        for j in range(len(board[i])):  # for each cell
            if board[i][j] == 0:
                print("   |", end="")  # empty
            elif board[i][j] == 1:
                print(" X |", end="")  # X
            elif board[i][j] == 2:
                print(" O |", end="")  # O
        print()

    print(" ---" * len(board[0]))  # bottom line
````

*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
#Write your solution here
def check_game(board):
    n = len(board)  # size of the board

    # check rows
    for i in range(n):
        same = True
        for j in range(1, n):
            if board[i][j] != board[i][0]:
                same = False
        if same and board[i][0] != 0:
            return board[i][0]  # winner in this row

    # check columns
    for j in range(n):
        same = True
        for i in range(1, n):
            if board[i][j] != board[0][j]:
                same = False
        if same and board[0][j] != 0:
            return board[0][j]  # winner in this column

    # check main diagonal
    same = True
    for i in range(1, n):
        if board[i][i] != board[0][0]:
            same = False
    if same and board[0][0] != 0:
        return board[0][0]  # winner in main diagonal

    # check anti-diagonal
    same = True
    for i in range(1, n):
        if board[i][n - 1 - i] != board[0][n - 1]:
            same = False
    if same and board[0][n - 1] != 0:
        return board[0][n - 1]  # winner in anti-diagonal

    # check if any empty space left
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                return -1  # game not finished

    return 0  # draw
````

````python
# Write your solution here
def place_move(board, player, row, col):
    # check if the position is inside the board
    if 0 <= row < len(board) and 0 <= col < len(board):
        # check if the cell is empty
        if board[row][col] == 0:
            board[row][col] = player  # store 1 for X, 2 for O
            return True  # move successful
    return False  # move not allowed
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
# Write your solution here
def play_game():
    board = create_board(3)  # create a 3x3 board
    player = 1  # start with player 1
    game_over = False  # track if the game has ended

    draw_tic_tac_toe(board)  # show the empty board

    while not game_over:
        player_turn(board, player)  # let the current player make a move

        result = check_game(board)  # check if someone has won or if it's a draw
        draw_tic_tac_toe(board)     # redraw the board after the move

        if result == 1:
            print("Player 1 wins!")  # player 1 wins
            game_over = True
        elif result == 2:
            print("Player 2 wins!")  # player 2 wins
            game_over = True
        elif result == 0:
            print("It's a draw!")  # no empty spaces left, it's a draw
            game_over = True
        else:
            player = 2 if player == 1 else 1  # switch to the other player
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
# create an empty n×n board
def create_board(n):
    board = []
    for i in range(n):
        row = []
        for j in range(n):
            row.append(0)  # 0 = empty cell
        board.append(row)
    return board

# draw the board with X, O, and row/column labels
def draw_tic_tac_toe(board):
    r = len(board)
    c = len(board[0])
    labels = ["A","B","C","D","E","F","G","H","I","J"]  # column labels

    # print column headers
    print("    ", end="")
    for j in range(c):
        print(labels[j], end="   ")
    print()

    # print each row with horizontal lines and cell contents
    for i in range(r):
        print("   ", end="")
        for j in range(c):
            print("---", end=" ")  # top line of cells
        print()
        print(str(i+1), end=" ")  # row number
        for j in range(c):
            print("|", end="")
            if board[i][j] == 0:
                print("   ", end="")  # empty cell
            elif board[i][j] == 1:
                print(" X ", end="")  # player 1
            elif board[i][j] == 2:
                print(" O ", end="")  # computer
        print("|")
    # bottom line of the last row
    print("   ", end="")
    for j in range(c):
        print("---", end=" ")
    print()

# check the current status of the game
def check_game(board):
    n = len(board)

    # check rows for winner
    for i in range(n):
        same = True
        for j in range(1, n):
            if board[i][j] != board[i][0]:
                same = False
        if same and board[i][0] != 0:
            return board[i][0]

    # check columns for winner
    for j in range(n):
        same = True
        for i in range(1, n):
            if board[i][j] != board[0][j]:
                same = False
        if same and board[0][j] != 0:
            return board[0][j]

    # check main diagonal
    same = True
    for i in range(1, n):
        if board[i][i] != board[0][0]:
            same = False
    if same and board[0][0] != 0:
        return board[0][0]

    # check anti-diagonal
    same = True
    for i in range(1, n):
        if board[i][n-1-i] != board[0][n-1]:
            same = False
    if same and board[0][n-1] != 0:
        return board[0][n-1]

    # check if any empty cells remain
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                return -1  # game not finished

    return 0  # draw

# place a move on the board
def place_move(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player  # mark the cell
        return True
    return False  # cell already occupied

# convert location string (e.g., "A1") to row/col and make a move
def make_move(board, player, location):
    labels = ["A","B","C","D","E","F","G","H","I","J"]
    col = -1
    # convert column letter to index
    for i in range(len(labels)):
        if location[0].upper() == labels[i]:
            col = i
            break
    if col == -1:
        return False
    try:
        row = int(location[1:]) - 1  # convert row number to index
    except:
        return False
    # check bounds
    if row < 0 or row >= len(board):
        return False
    if col < 0 or col >= len(board[0]):
        return False
    return place_move(board, player, row, col)

# handle human player's turn
def player_turn(board, player):
    move_ok = False
    while not move_ok:
        location = input("Enter a location (e.g., A1): ")
        move_ok = make_move(board, player, location)
        if not move_ok:
            print("Invalid move. Try again.")

# basic AI logic for the computer
def computer_move(board):
    n = len(board)

    # try to win
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                board[i][j] = 2
                if check_game(board) == 2:
                    return
                board[i][j] = 0

    # block player's winning move
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                board[i][j] = 1
                if check_game(board) == 1:
                    board[i][j] = 2
                    return
                board[i][j] = 0

    # otherwise, pick first empty cell
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                board[i][j] = 2
                return

# main game loop for player vs computer
def play_game_vs_computer():
    size = int(input("Enter board size n: "))
    board = create_board(size)
    player = 1  # human starts
    game_over = False

    while not game_over:
        draw_tic_tac_toe(board)  # show current board
        if player == 1:
            print("Player 1's turn:")
            player_turn(board, player)
        else:
            print("Computer's turn:")
            computer_move(board)

        result = check_game(board)
        if result == 1:
            draw_tic_tac_toe(board)
            print("Player 1 wins!")
            game_over = True
        elif result == 2:
            draw_tic_tac_toe(board)
            print("Computer wins!")
            game_over = True
        elif result == 0:
            draw_tic_tac_toe(board)
            print("It's a draw!")
            game_over = True
        else:
            player = 2 if player == 1 else 1  # switch turns
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
def evaluate(board):
    # return a score for the board: +100 if computer wins, -100 if player wins, 0 otherwise
    result = check_game(board)
    if result == 2:
        return 100
    elif result == 1:
        return -100
    else:
        return 0

def minimax(board, depth, max_depth, is_computer):
    # evaluate current board
    score = evaluate(board)
    # stop recursion if someone won, draw, or reached max depth
    if score == 100 or score == -100 or check_game(board) == 0 or depth == max_depth:
        return score

    n = len(board)

    if is_computer:
        best = -100
        for i in range(n):
            for j in range(n):
                if board[i][j] == 0:
                    board[i][j] = 2  # try computer move
                    val = minimax(board, depth + 1, max_depth, False)
                    if val > best:
                        best = val
                    board[i][j] = 0  # undo move
        return best
    else:
        best = 100
        for i in range(n):
            for j in range(n):
                if board[i][j] == 0:
                    board[i][j] = 1  # try player move
                    val = minimax(board, depth + 1, max_depth, True)
                    if val < best:
                        best = val
                    board[i][j] = 0  # undo move
        return best

def computer_move_minimax(board, max_depth=3):
    best_val = -100
    best_move = None
    n = len(board)

    # try all moves and pick the one with the best minimax value
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                board[i][j] = 2
                val = minimax(board, 0, max_depth, False)
                board[i][j] = 0
                if val > best_val:
                    best_val = val
                    best_move = (i, j)

    # fallback: pick first empty cell if no best move found
    if best_move is None:
        for i in range(n):
            for j in range(n):
                if board[i][j] == 0:
                    best_move = (i, j)
                    break
            if best_move is not None:
                break

    # make the move
    if best_move:
        board[best_move[0]][best_move[1]] = 2
        return True
    return False

def play_game_vs_computer_minimax():
    n = int(input("Enter board size n: "))
    board = create_board(n)
    player = 1
    game_over = False

    draw_tic_tac_toe(board)

    while not game_over:
        if player == 1:
            print("Player 1's turn:")
            player_turn(board, player)
        else:
            print("Computer's turn:")
            computer_move_minimax(board)

        draw_tic_tac_toe(board)
        result = check_game(board)

        if result == 1:
            print("Player 1 wins!")
            game_over = True
        elif result == 2:
            print("Computer wins!")
            game_over = True
        elif result == 0:
            print("It's a draw!")
            game_over = True
        else:
            player = 2 if player == 1 else 1  # switch player
````

````python
# Computer player 1: chooses first empty cell
def comp_player_1_move(board, player):
    n = len(board)
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                board[i][j] = player  # place move
                return True
    return False

# Computer player 2: smarter AI chooses best move
def comp_player_2_move(board, player, max_depth):
    best_score = -100
    best_move = None
    n = len(board)
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                board[i][j] = player  # try move
                score = evaluate_move(board, 0, max_depth, False, player)  # get score
                board[i][j] = 0  # undo move
                if score > best_score:
                    best_score = score
                    best_move = (i, j)
    if best_move:
        board[best_move[0]][best_move[1]] = player  # make best move
        return True
    return False

# Function to give a score for the board
def evaluate_move(board, depth, max_depth, is_player_turn, player_number):
    result = check_game(board)
    if result == player_number:  # current player wins
        return 100
    elif result != -1 and result != player_number:  # other player wins
        return -100
    elif result == 0 or depth == max_depth:  # draw or depth limit
        return 0

    n = len(board)
    if is_player_turn:
        best = -100
        for i in range(n):
            for j in range(n):
                if board[i][j] == 0:
                    board[i][j] = player_number
                    best = max(best, evaluate_move(board, depth + 1, max_depth, False, player_number))
                    board[i][j] = 0  # undo move
        return best
    else:
        opponent = 1 if player_number == 2 else 2
        best = 100
        for i in range(n):
            for j in range(n):
                if board[i][j] == 0:
                    board[i][j] = opponent
                    best = min(best, evaluate_move(board, depth + 1, max_depth, True, player_number))
                    board[i][j] = 0  # undo move
        return best

# Simulate one game between the two computer players
def simulate_game(board_size, max_depth):
    board = create_board(board_size)
    player = 1
    while True:
        if player == 1:
            comp_player_1_move(board, player)  # player 1 move
        else:
            comp_player_2_move(board, player, max_depth)  # player 2 move
        result = check_game(board)
        if result != -1:  # game over
            return result
        player = 2 if player == 1 else 1  # switch player

# Run multiple games for different board sizes
def run_simulations():
    sizes = [3, 4, 5]  # board sizes
    max_depths = {3: 6, 4: 4, 5: 3}  # depth limit for smarter player
    games_per_size = 10  # number of games

    for size in sizes:
        wins_comp_1 = 0
        wins_comp_2 = 0
        draws = 0
        for _ in range(games_per_size):
            result = simulate_game(size, max_depths[size])
            if result == 1:
                wins_comp_1 += 1
            elif result == 2:
                wins_comp_2 += 1
            else:
                draws += 1
        # print results
        print(f"{size}x{size} board:")
        print(f"Comp Player 1 wins: {wins_comp_1}")
        print(f"Comp Player 2 wins: {wins_comp_2}")
        print(f"Draws: {draws}")
        win_rate = wins_comp_2 / games_per_size * 100
        print(f"Comp Player 2 win rate: {win_rate}%\n")
````
